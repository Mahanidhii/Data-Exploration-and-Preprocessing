### Date Time Operations

```python
df["pickup"] = pd.to_datetime(df["pickup"])
df["dropoff"] = pd.to_datetime(df["dropoff"])
```

1. Trips per hour

```python
df["pickup_hour"] = df["pickup"].dt.hour
hour_counts = df["pickup_hour"].value_counts().sort_index()

plt.figure()
plt.bar(hour_counts.index, hour_counts.values)
plt.xlabel("Hour of the Day")
plt.ylabel("Number of Trips")
plt.title("No. of trips by Pickup Hour")
plt.show()
```

2. Trips per Month

```python
df["pickup_month"] = df["pickup"].dt.month_name()

month_order = ["January","February","March","April","May","June",
               "July","August","September","October","November","December"]

df["pickup_month"].value_counts().reindex(month_order)
```

3. Trips per weekday

```python
df["pickup_weekday"] = df["pickup"].dt.day_name()
df["pickup_weekday"].value_counts()
```

4. Average trips per hour

```python
print(df.groupby(df["pickup"].dt.hour).size().mean())
```

5. Resampling daily total fare

```python
df_time = df.set_index("pickup")

df_time.resample("D")["fare"].sum().head()
```

6. Average trip duration per hour

```python
print((df["dropoff"] - df["pickup"]).dt.total_seconds().mean() / 60)
```

7. Trips per hour

```python
df.groupby("pickup_hour")["fare"].count()
```

8. Resampling daily total fare

```python
df.set_index("pickup", inplace=True)
df.resample("D")["fare"].sum()
```
